```python
import pandas as pd

df = pd.read_csv('/home/dean/Desktop/Lecture/Machine Learning/Codes/Data.csv')
print (df.isnull().sum())
df['Age'] = df['Age'].fillna(df['Age'].mean().round(2))
df['Salary'] = df['Salary'].fillna(df['Salary'].mean().round(2))
df['Country'] = df['Country'].ffill()
print (df.isnull().sum())
print (df)
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer as sp

df = pd.read_csv('/home/dean/Desktop/Lecture/Machine Learning/Codes/Data.csv')
print (df.isnull().sum())

imputer = sp(strategy = 'most_frequent')
df_impute = pd.DataFrame(imputer.fit_transform(df), columns = df.columns)
print (df_impute)
```
