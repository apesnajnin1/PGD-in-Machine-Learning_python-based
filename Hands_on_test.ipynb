````python


# Ans 1.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


#Ans 2.
df = pd.read_excel("yahoo_data.xlsx")

print(df.head())
print(df.info())
print(df.isnull().sum())


#Ans 3.


df['Next_Day_Close'] = df['Adj Close'].shift(-1)


df['Lag_1'] = df['Adj Close'].shift(1)


df['Daily_Return'] = (
    (df['Adj Close'] - df['Lag_1']) / df['Lag_1']
)


df['Volatility'] = (
    (df['High'] - df['Low']) / df['Open']
)


df['Volume_Lag_1'] = df['Volume'].shift(1)


# Ans 4.
df = df.dropna()

print("After cleaning:")
print(df.isnull().sum())


# Ans 5.
X = df[['Lag_1',
        'Daily_Return',
        'Volatility',
        'Volume_Lag_1']]

y = df['Next_Day_Close']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    shuffle=False
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)


# Ans 6

models = {
    "Least Squares": LinearRegression(),
    "Ridge": Ridge(alpha=1.0),
    "Lasso": Lasso(alpha=0.001, max_iter=10000)
}

results = []

for name, model in models.items():

    # Fit model
    model.fit(X_train, y_train)

    # Predictions
    y_pred = model.predict(X_test)

    # Metrics
    mae = mean_absolute_error(y_test, y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_test, y_pred)
    )

    r2 = r2_score(y_test, y_pred)

    results.append([
        name,
        mae,
        rmse,
        r2
    ])

# Ans 7

results_df = pd.DataFrame(
    results,
    columns=['Model', 'MAE', 'RMSE', 'R-Squared']
)

print("\nModel Comparison:")
print(results_df)



best_model = results_df.loc[
    results_df['R-Squared'].idxmax()
]

print("\nBest Model:")
print(best_model)



plt.figure(figsize=(10, 5))
plt.plot(y_test.values, label='Actual')
plt.plot(
    models['Least Squares'].predict(X_test),
    label='Predicted'
)

plt.title('Actual vs Predicted Next Day Close')
plt.xlabel('Test Samples')
plt.ylabel('Adjusted Close')
plt.legend()
plt.show()
````
